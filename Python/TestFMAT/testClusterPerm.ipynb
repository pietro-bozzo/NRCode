```python
%load_ext autoreload
%autoreload 2
import numpy as np
import scipy as sp
import fmatoolbox as fma
```

```python
x = np.random.randn(20,50)
y = np.random.randn(20,50) + 2 * sp.stats.Normal().pdf(np.linspace(-10,10,50))[None,:]
stat = fma.analysis.clusterPermutationTest(x,y,paired=True)
print(stat['sig_mask'])
```

```python
fig, ax = fma.plotting.makeFigure(size=(12,5))
ax[0].plot(x.mean(0))
ax[0].plot(y.mean(0))
ax[0].plot((x-y).mean(0))
fma.plotting.pHorzLine(stat['sig_mask'])
```

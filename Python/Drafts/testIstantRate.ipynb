```python
%load_ext autoreload
%autoreload 2
import numpy as np
import fmatoolbox as fma
batch_file = '/mnt/hubel-data-103/Pietro/InfraSlowNRPaper/Data/IS_intervals.batch'
```

```python
session = fma.data.readBatchFile(batch_file)[0][15]
print(session)
R = fma.regions.regions(session)
spikes = R.spikes()
```

```python
bin = 0.1
rate1 = fma.analysis.istantaneousRate(spikes[:,:],bin=bin,step=5)
t1 = rate1[:,0]
rate1 = rate1[:,1:] * bin
```

```python
rate2 = fma.analysis.istantaneousRateNEW(spikes[:,:],bin=bin,step=5)
t2 = rate2[:,0]
rate2 = rate2[:,1:] * bin
```

```python
A = np.where(~np.isclose(rate1,rate2))
print(np.all(np.isclose(t1,t2)))
print(np.sum(np.abs(rate1-rate2)) / len(spikes))
print(np.max(np.abs(rate1-rate2)))
```

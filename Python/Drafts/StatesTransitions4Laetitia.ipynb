```python
%load_ext autoreload
%autoreload 2
import numpy as np
import fmatoolbox as fma
batch_file = '/mnt/hubel-data-103/Pietro/InfraSlowNRPaper/Data/IS_intervals.batch'
do_save = False
```

```python
session = fma.data.readBatchFile(batch_file)[0][12]
#session = '/mnt/hubel-data-149/Rat012/Rat012_2025-12-17/Rat012_2025-12-17.xml'
print(session)
R = fma.regions.regions(session,phases='sleep.*',states=['sws','rem'])
```

```python
recording = R.eventIntervals()
sws = R.eventIntervals('sws')
rem = R.eventIntervals('rem')

print(np.diff(recording).sum() / 3600)
print(len(sws),len(rem))

times_a = sws[:,1]
times_b = rem[:,0]+0.001
# smallest times_b following each event a
idx = np.searchsorted(times_b,times_a,side='right')
valid = idx < len(times_b)
distance = np.full_like(times_a,np.nan,dtype=float)
distance[valid] = times_b[idx[valid]] - times_a[valid]
is_coupled_a = (distance >= 0) & (distance <= 0.01)
print(is_coupled_a.sum())
```

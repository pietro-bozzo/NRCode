```python
import numpy as np
from scipy.signal import butter, filtfilt, find_peaks

def population_rate_from_spikes(spike_times_list, T, bin_size=0.01):
    edges = np.arange(0, T + bin_size, bin_size)
    pop_counts = np.zeros(len(edges) - 1)
    for st in spike_times_list:
        counts, _ = np.histogram(st, bins=edges)
        pop_counts += counts
    rate = pop_counts / (bin_size * len(spike_times_list))  # Hz par neurone
    t = edges[:-1] + bin_size/2
    return t, rate

def bandpass(x, fs, low, high, order=4):
    from scipy.signal import butter, filtfilt
    nyq = 0.5 * fs
    b, a = butter(order, [low/nyq, high/nyq], btype='band')
    return filtfilt(b, a, x)

def delta_episodes_from_spikes(t, rate,
                               bin_size=0.01, low=0.5, high=4.0,
                               amp_thresh_factor=1.5, min_interval_s=0.25):
    # 1. Construire le firing rate de population

    fs = 1.0 / bin_size

    # 2. Filtre delta
    rate_delta = bandpass(rate, fs, low, high)    ################## Pas forcément pertinent 

    # 3. Seuil d'amplitude
    amp_thresh = amp_thresh_factor * np.median(np.abs(rate_delta))
    min_samples = int(min_interval_s * fs)

    # 4. Détection de pics (facultatif mais on les garde pour info)
    peaks, _ = find_peaks(-rate_delta, height=amp_thresh, distance=min_samples) # -rate pour avoir les peaks d'inactivité     
    t_peaks = t[peaks]

    # 5. Détection des épisodes (intervalle où rate_delta > seuil)
    above = rate_delta > amp_thresh
    ######################################################################################################################
    silence_threshold = np.percentile(rate, 20)  # 20% des valeurs les plus basses
    above = rate_delta <= silence_threshold
    ######################################################################################################################

    # transitions 0->1 et 1->0
    diff = np.diff(above.astype(int))
    starts = np.where(diff == 1)[0] + 1
    ends   = np.where(diff == -1)[0] + 1

    # cas où on commence déjà au-dessus ou on finit au-dessus
    if above[0]:
        starts = np.insert(starts, 0, 0)
    if above[-1]:
        ends = np.append(ends, len(above) - 1)

    # les épisodes de silence sont les périodes où rate <= silence_threshold

    # Construire la liste des [t_deb, t_fin]
    episodes = []
    for s, e in zip(starts, ends):
        # on peut imposer une durée minimale (en échantillons)
        if e > s:
            episodes.append([t[s], t[e]])

    episodes = np.array(episodes) if len(episodes) > 0 else np.empty((0, 2))

    # intervals entre épisodes = entre t_deb successifs par exemple
    episode_intervals = np.diff(episodes[:, 0]) if len(episodes) > 1 else np.array([])

    return rate_delta, episodes, episode_intervals, t_peaks
```

```python
import matplotlib.pyplot as plt

def plot_rate_with_delta_episodes(t, rate, episodes, tdeb=0, tmax=None,
                                  rate_delta=None,
                                  show_delta=True,
                                  alpha_fill=0.2):
    """
    t : vecteur temps
    rate : firing rate (Hz)
    episodes : array shape (N, 2) avec [t_deb, t_fin] pour chaque épisode
    rate_delta : (optionnel) rate filtré delta, même taille que rate
    show_delta : si True, trace aussi rate_delta
    """
    fig, ax = plt.subplots(figsize=(10, 4))

    t = np.array(t)
    ind = t <= tmax
    t = t[ind] 
    rate = rate[ind]
    rate_delta = rate_delta[ind]
    episodes = episodes[episodes[:, 1]<=tmax] 

    ax.plot(t, rate, color='black', lw=1, label='Firing rate')

    if show_delta and rate_delta is not None:
        ax.plot(t, rate_delta, color='C1', lw=1, label='Delta-filtered rate')

    # Colorier les épisodes
    for (t_start, t_end) in episodes:
        ax.axvspan(t_start, t_end, color='C0', alpha=alpha_fill)

    ax.set_xlabel('Time (s)')
    ax.set_ylabel('Rate (Hz)')
    ax.set_title('Firing rate with delta-like episodes')
    ax.legend(loc='upper right')
    fig.tight_layout()
    return fig, ax

 

 

rate_delta, episodes, episode_intervals, t_peaks = delta_episodes_from_spikes(t, rate, bin_size=0.01, low=0.5, high=4.0)

fig, ax = plot_rate_with_delta_episodes(t, rate, episodes, tmax=270, rate_delta=rate_delta)
plt.show()
```

```python
def avalanche_score(session, bin_size, threshold):
    spikes, FR, size, intervals, size_t = avalanchesresult(session, neurons=15, tmax=15, smoothed=False, bin_size=bin_size, threshold=threshold)
    if len(size) < 2:
        return -np.inf

    size = np.array(size)

    n_aval = len(size)
    var_size = np.var(size)

    total_time = FR.shape[0] * bin_size
    active_time = sum([t2-t1 for t1,t2 in intervals])
    frac_active = active_time / total_time

    # pénalités
    score = (
        0.0001 * var_size
        - 12.0 * frac_active
        - 0.4 * n_aval
    )

    print(f"Score: {score:.2f} (var_size={var_size:.2f}, frac_active={frac_active:.2f}, n_aval={n_aval})")

    return score
```

```python
def avalanche_score_V2(session, bin_size, threshold, spikes, spikes_indiv, tmax=10000, w1=0.8, w2=2.0, w3=0.1):
    spikes, FR, size, intervals, size_t = avalanchesresult_V2(session, spikes, spikes_indiv, tmax=tmax, bin_size=bin_size, threshold=threshold)
    if len(size) < 2:
        return -np.inf

    size = np.array(size)

    n_aval = len(size)
    var_size = np.var(size)

    total_time = FR.shape[0] * bin_size
    active_time = sum([t2-t1 for t1,t2 in intervals])
    frac_active = active_time / total_time

    p1 = (var_size / np.mean(size)**2)
    p3 = (n_aval / (total_time/np.mean([t2-t1 for t1,t2 in intervals])))

    # pénalités
    score = (
    w1 * p1
    - w2 * frac_active
    - w3 * p3
    )

    print(f"Score: {score:.2f} (var_size={p1:.2f}, frac_active={frac_active:.2f}, n_aval={p3})")

    return score
```

```python
import powerlaw
from scipy.stats import linregress


def compute_dcc(sizes, durations):

    sizes = np.array(sizes)
    durations = np.array(durations)

    # --- exponent tau (size distribution)
    fit_s = powerlaw.Fit(sizes, discrete=True, verbose=False)
    tau = fit_s.power_law.alpha

    # --- exponent alpha (duration distribution)
    fit_t = powerlaw.Fit(durations, discrete=True, verbose=False)
    alpha = fit_t.power_law.alpha

    # --- gamma measured
    unique_T = np.unique(durations)

    mean_sizes = []
    for T in unique_T:
        mean_sizes.append(np.mean(sizes[durations == T]))

    mean_sizes = np.array(mean_sizes)

    slope, _, _, _, _ = linregress(
        np.log(unique_T),
        np.log(mean_sizes)
    )

    gamma_measured = slope

    # --- predicted gamma
    gamma_pred = (alpha - 1) / (tau - 1)

    # --- DCC
    dcc = abs(gamma_measured - gamma_pred)

    return dcc, tau, alpha, gamma_measured
```

```python
def avalanche_loss(session, bin_size, threshold):

    spikes, FR, sizes, intervals, size_t = avalanchesresult(session, neurons=15, tmax=15, smoothed=False, bin_size=bin_size, threshold=threshold)
    durations = np.array(intervals[:,1])-np.array(intervals[:,0])

    if len(sizes) < 2:
        return np.inf, np.inf

    dcc, tau, alpha, gamma = compute_dcc(sizes, durations)

    return dcc, gamma
```

```python
def optimize_bin_size_powerlaw(session, threshold, b0=0.2, lr=0.01, eps=1e-3, steps=20):

    b = b0

    for i in range(steps):
        s_plus, gp = avalanche_loss(session, b + eps, threshold)
        s_minus, gm = avalanche_loss(session, b - eps, threshold)
        print(s_plus, s_minus, gp, gm)

        grad = (s_plus - s_minus) / (2 * eps)

        b = b - lr * grad

        print(f"step {i}: bin_size={b:.4f}, grad={grad:.4f} -> loss={avalanche_loss(session,b,threshold)}, score={avalanche_score(session,b,threshold)}")

    return b
```

```python
def avalanchesLitteratureplot(session, FR=None, bin_size=0.00295, neurons=None, tmax=50, threshold=None, smoothed=False, window=5):
    
    spikes_indiv = fma.data.loadSpikeTimes(session)
    if neurons is not None:
        spikes_indiv = dict(list(spikes_indiv.items())[:neurons])
    spikes_indiv = {key: fma.general.restrict(spikes_indiv[key], [0, tmax]) for key in spikes_indiv} 
    
    spikes = fma.data.loadSpikeTimes(session, output='compact')
    if neurons is not None:
        ind = spikes[:,1] < neurons
        spikes = spikes[ind,:]
    spikes = spikes[(spikes[:,0] >= 0) & (spikes[:,0] <= tmax),:]
    
    if FR is None:
        if threshold is None:
            bin_size = 0.00295  # Δt = 2.95 msec
        FR = fma.analysis.firingRate(spikes[:,0], bin_size=bin_size)
    
    time = FR[:, 0]
    rates = FR[:, 1]
    
    # déterminer le threshold
    if threshold is None:
        mean = np.mean(rates)
        sd   = np.std(rates)
        threshold_val = mean + 3 * sd  # +3 sd
        # convertir en percentile pour avalanchesFromProfile
        threshold = np.sum(rates < threshold_val) / len(rates) * 100
        print(f"Auto threshold: mean={mean:.2f}, sd={sd:.2f}, +3sd={threshold_val:.2f} -> percentile={threshold:.2f}")
    
    if smoothed:
        smoothed_rates = np.convolve(rates, np.ones(window)/window, mode='same')
    
    size, intervals, size_t = rg.computation.avalanchesFromProfile(
        rates,
        threshold=threshold,
        time_step=FR[1,0] - FR[0,0],
        t0=FR[0,0]
    )
    
    # ... reste inchangé

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(20,10), sharex=True)

    for n, (neuron_id, spike_times) in enumerate(spikes_indiv.items()):
    # transforme la liste en array si nécessaire
        spike_times = np.array(spike_times)
    
    # y = n pour ce neurone
        ax1.scatter(spike_times, 
                [neuron_id] * len(spike_times), 
                marker='|', 
                s=100,     # taille du trait
                color='black')
    
    for t1, t2 in intervals:
        ax1.axvspan(t1, t2, color='red', alpha=0.3)


    ax2.plot(time[:len(rates)] , rates, color='green', label='Firing rate')
    if smoothed:
        ax2.plot(time[:len(smoothed_rates)], smoothed_rates, color='blue', label='Smoothed firing rate')
    for t1, t2 in intervals:
        ax2.axvspan(t1, t2, color='red', alpha=0.3)

    ax1.set_xlabel("Time (s)")
    ax1.set_ylabel("Neuron Index")
    ax1.set_title("Spikes with avalanches")

    ax2.set_xlabel("Time (s)")
    ax2.set_ylabel("Firing rate (Hz)")
    ax2.set_title("Firing rate (moving average) with avalanches")
    ax2.legend()

    
    plt.show()
```
